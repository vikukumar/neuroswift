## 1. Setup & Imports

```python
import torch
import neuroswift
from neuroswift import NeuroSwiftLM, WordTokenizer, auto_device
from pathlib import Path

device = auto_device()
print(f"Using device: {device}")
```

## 2. Ultra-Fast CPU Optimization

NeuroSwift uses vectorized associative scans for God-level throughput on CPU. You can further optimize with `model.compile()`.

```python
from neuroswift import NeuroSwiftConfig

# Minimal config for demonstration
config = NeuroSwiftConfig(vocab_size=1000, d_model=128, n_layers=2)
model = NeuroSwiftLM(config).to(device)

# Enable JIT compilation for massive speedups (requires MSVC/GCC)
model = model.compile(mode="reduce-overhead")
```
